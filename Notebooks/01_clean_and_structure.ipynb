```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_csv('../Data/marketing_campaign.csv', sep=';')
df.head()
```

```python
duplicate_count = df.duplicated().sum()
print(f'Total Duplicate Rows Found: {duplicate_count}')
if duplicate_count > 0:
    df = df.drop_duplicates()
    print('Duplicate Rows Removed Successfully!')
else:
        print('No Duplicate Rows Found.')
print(f'Current Shape of Dataset: {df.shape}')
```

```python
df.info()
df['Dt_Customer'] = pd.to_datetime(df['Dt_Customer'], format='%Y-%m-%d')
print("\n new data type of 'Dt_Customer':", df['Dt_Customer'].dtype)
df[['Dt_Customer']].head()
```

```python
print("Missing values in Income before fix:", df['Income'].isnull().sum())
df['Income'] = df['Income'].fillna(df['Income'].median())
df = df[df['Year_Birth'] > 1900]
df = df[df['Income'] < 200000]
print("Missing values in Income after fix:", df['Income'].isnull().sum())
print(f"Data shape after removing anomalies: {df.shape}")
```

```python
df['Marital_Status'] = df['Marital_Status'].replace({
'Alone': 'Single',
'Absurd': 'Single',
'YOLO': 'Single',
'Together': 'Married'
})
df['Education'] = df['Education'].str.strip()
df['Marital_Status'] = df['Marital_Status'].str.strip()
print("Cleaned Marital Status Categories:")
print(df['Marital_Status'].value_counts())
```

```python
df.to_csv('../Data/cleaned_data.csv', index=False)
print('Cleaned data successfully saved to "../Data/cleaned_data.csv" !')
```
