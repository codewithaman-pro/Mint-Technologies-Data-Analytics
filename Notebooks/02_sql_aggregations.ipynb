```python
import pandas as pd
import sqlite3
df = pd.read_csv('../Data/cleaned_data.csv')
conn = sqlite3.connect(':memory:')
df.to_sql('marketing_data', conn, index=False, if_exists='replace')
print("SQL Database setup successfully! Table 'marketing_data' created.")
```

```python
query1 = """
SELECT 
    Education,
    COUNT(*) AS Total_Customers,
    ROUND(AVG(Income), 2) AS Avg_Income,
    ROUND(AVG(MntWines + MntFruits + MntMeatProducts + MntFishProducts + MntSweetProducts + MntGoldProds), 2) AS Avg_Total_Spending
FROM marketing_data
GROUP BY Education
ORDER BY Avg_Total_Spending DESC;
"""

df_education_summary = pd.read_sql_query(query1, conn)
print("--- Education Level Aggregation ---")
print(df_education_summary)

print("\n" + "="*50 + "\n")

query2 = """
SELECT 
    Marital_Status,
    COUNT(*) AS Total_Customers,
    SUM(AcceptedCmp1 + AcceptedCmp2 + AcceptedCmp3 + AcceptedCmp4 + AcceptedCmp5) AS Total_Campaigns_Accepted,
    ROUND(AVG(NumWebPurchases + NumCatalogPurchases + NumStorePurchases), 2) AS Avg_Total_Purchases
FROM marketing_data
GROUP BY Marital_Status
ORDER BY Total_Campaigns_Accepted DESC;
"""

df_marital_summary = pd.read_sql_query(query2, conn)
print("--- Marital Status Aggregation ---")
print(df_marital_summary)
```
